## First, grab some data and do a calculation on it

We want to see that we can display calculated data from an in-memory dataset.

### Get Sentinel 2 data as an Xarray `DataSet`

#### Define a Pyrenees Bounding Box

````python
from leafmap.plot import bbox_to_gdf

bbox = [-1.7, 42.2, 3.0, 43.4]
bbox_to_gdf(bbox).explore()
````

#### Query the Element84 Earth Search STAC catalog for Sentinel2 data

````python
import pystac_client
import planetary_computer

catalog = pystac_client.Client.open(
    "https://planetarycomputer.microsoft.com/api/stac/v1",
    modifier=planetary_computer.sign_inplace,
)
````

````python
from pystac_client import Client

sentinel2_stac_items = (
    catalog
    .search(
        collections=["sentinel-2-l2a"],
        bbox=bbox,
        datetime="2024-04-25/2024-06-25",
        query={"eo:cloud_cover": {"lt": 15}},
    )
    .item_collection()
)
len(sentinel2_stac_items)
````

#### Load the data into an Xarray `Dataset`

````python
import odc.stac

sentinel2_dataset = odc.stac.load(
    sentinel2_stac_items,
    bands=["red", "green", "blue", "swir16"],
    chunks={'x': 2048, 'y': 2048},
    bbox=bbox,
    patch_url=planetary_computer.sign,
    resolution=20,
    groupby="solar_day",
    nodata=0,
)
sentinel2_dataset
````

````python
config_options = {
                  'AWS_VIRTUAL_HOSTING': 'FALSE',
                  'GDAL_HTTP_UNSAFESSL': 'YES',
                  'AWS_HTTPS': 'YES',
                  'AWS_NO_SIGN_REQUEST': 'YES'}
import rasterio
````

````python
from dask_kubernetes.operator import KubeCluster

config = {
   "name": "ndsi-jgis",
   "namespace": "jhub",
   "image": "ghcr.io/geojupyter/presentation-cng2026:sha-723a7c3",
   "n_workers": 2,
   "resources":{"requests": {"memory": "4Gi"}, "limits": {"memory": "4Gi"}}
}

cluster = KubeCluster(**config)
````

````python
cluster.scale(20)
cluster
````

````python
dask_client = cluster.get_client()
dask_client
````

````python
def configure_workers_environment():
    import os
    for key, value in config_options.items():
        os.environ[key] = value
        
    rio_env = rasterio.Env(**config_options)
    rio_env.__enter__()
    return None
    
dask_client.register_worker_callbacks(configure_workers_environment)
````

### Visualize raw data in RGB

````python
%%time
with rasterio.Env(**config_options):
    rgb = sentinel2_dataset[["red", "green", "blue"]].to_array(dim="band").max("time")
    rgb_scaled = (rgb / 3000).clip(0, 1)  # Scale and clip for display
    rgb_scaled
````

````python
#Visualize subset
with rasterio.Env(**config_options):
    rgb_scaled[:,5000:6000,10000:12000].plot.imshow()
````

### Calculate NDSI

````python
%%time
with rasterio.Env(**config_options):
    # Filter on red band value
    sentinel2_dataset_red_high = sentinel2_dataset.where(sentinel2_dataset.red.mean(dim="time") > 200)
    
    ndsi = (
        (
            (sentinel2_dataset_red_high.green - sentinel2_dataset_red_high.swir16)
            / (sentinel2_dataset_red_high.green + sentinel2_dataset_red_high.swir16)
        )
        .max(dim="time")
        .where(lambda ndsi: ndsi < 1)
    )

    # Preload everything, avoid multiple get and bandwidth overload
    ndsi = ndsi.persist()
    
````

````python
ndsi[5000:6000,10000:12000].plot.imshow()
````

### Build a temporal STAC

````python
def get_sentinel2_stack(datetime="2022-04-25/2022-07-01"):
    sentinel2_stac_items = (
        catalog
        .search(
            collections=["sentinel-2-l2a"],
            bbox=bbox,
            datetime=datetime,
            query={"eo:cloud_cover": {"lt": 22}}, #Cloudy month? But too much cloud is wrong
        )
        .item_collection()
    )
    print(len(sentinel2_stac_items))

    sentinel2_dataset = odc.stac.load(
        sentinel2_stac_items,
        bands=["red", "green", "blue", "swir16"],
        chunks={'x': 2048, 'y': 2048},
        bbox=bbox,
        patch_url=planetary_computer.sign,
        resolution=20,
        groupby="solar_day",
        nodata=0,
    )
    return sentinel2_dataset
    
````

````python
def compute_snow_bool(sentinel2_dataset):
    with rasterio.Env(**config_options):
    # Filter on red band value
        sentinel2_dataset_red_high = sentinel2_dataset.where(sentinel2_dataset.red.mean(dim="time") > 200)
        ndsi = (
            (
                (sentinel2_dataset_red_high.green - sentinel2_dataset_red_high.swir16)
                / (sentinel2_dataset_red_high.green + sentinel2_dataset_red_high.swir16)
            )
            .max(dim="time")
            .where(lambda ndsi: ndsi < 1)
        )
        snow = ndsi > 0.4
        snow = snow.astype('uint8')
    
        # Preload everything, avoid multiple get and bandwidth overload
        snow = snow.persist()
        return snow
````

````python
import xarray
import pandas

years = range(2017, 2025)
pindex = pandas.Index(data=years, name="time")
snows = []
for year in years:
    print(f'Creating datarray for year {year}')
    sentinel2_dataset = get_sentinel2_stack(datetime=f'{year}-04-20/{year}-06-20')
    snow_array = compute_snow_bool(sentinel2_dataset)
    snows.append(snow_array)

#snow_cube = xarray.concat(snows, pindex, coords="all")#, compat="override")
#snow_cube
````

````python
from rio_tiler.colormap import cmap
cmap_to_choose = cmap.list().copy()
import random

for index in range(len(years)):
    cmap_index = random.randint(1, len(cmap_to_choose))
    random_cmap = cmap_to_choose[cmap_index]
    cmap_to_choose.remove(random_cmap)
    
    await doc.add_data_array_layer(
        name=f"Snow Cover {years[index]}",
        data_array=snows[index],
        colormap_range=(0, 1),
        colormap_name=random_cmap
    )
````

### Show histogram

````python
snow_surfaces = []
pindex = pandas.Index(data=years, name="time")
for index in range(len(snows)):
    surface = snows[index].sum().compute() * 20*20 / (1000*1000)
    snow_surfaces.append(surface)

snow_surfaces_array = xarray.concat(snow_surfaces, pindex)
snow_surfaces_array
````
